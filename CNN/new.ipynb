```python
IMG_SIZE = 160

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weight='imagent'
)
```

```python
base_model.trainable = False
```

```python
global_average_layer = tf.keras_layers.GlobalAveragePooling2D()
```

```python
prediction_layer = tf.keras.layers.Dense(1)
```

```python
model = tf.keras
```

```python
input_shape = (160, 160 3)   # example input shape, adjust as needed

model = tf.keras.Sequential([
    tf.keras.layers.InputLayer(input_shape=input_shape), # add an input lauer
    base_model,
    global_average_layer,
    prediction_layer
])
```

```python
model.summary()
```

```python

base_learning_rate = 0.0001

model.compile(optimizer = tf.keras.optimizers.RMSprop(lr=base_learning_rate),
              loss = tf.keras.losses.BinaryCrossentropy(from_logits=True),
              metrics=['accuracy']
)

BATCH_SIZE = 32
SHUFFLE_BUFFER_SIZE = 1000

train_batches = train.shuffle(SHUFFLE_BUFFER_SIZE).batch(BATCH_SIZE)
validation_batches = validation.batch(BATCH_SIZE)
test_batches = test.batch(BATCH_SIZE)

initial_epochs = 3
validation_steps = 1000
loss, accuracy = model.evaluate(validation_batches, steps=validation_steps)
```
