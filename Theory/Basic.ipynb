# DEEP LEARNING


Deep Learning is a part of Machine Learning  thar uses 
artificial neural networks with multiple layers to learn 
pattern from large amounts of data
inspired by the human brain

Input  ----> Hidden layers -----> output

example  => face recognition, self-driving cars, voice assistants

# Neuron
Neuron is a methematical unit that takes input values, perform calculations using weight and bias and applies activation function and produces an output

        z = wx + b
        y = f(z)

- x  => input
- w  => weight
- b  => bias
- f(z) => activation function
- y   => output


# Layer
A Layer is a group of neurons in a neural network

# Neural Network
A system of multiple connected layers that learn pattern from data

- Weight  --> A numerical value that determines the importance of an 
               input in a calculation
- Bias  ---> A numerical value added to a calculation to adjust the output

# Types of neural networks:
- ANN: Artificial Neural Network.
- CNN: Convolutional Neural Network, commonly used for images.
- RNN: Recurrent Neural Network, designed for sequential data.
- LSTM: A type of RNN that can learn longer-term dependencies.
- Transformer: Widely used in language models and generative AI.
- Autoencoder: Learns compressed representations of data.

# Advantages:
- Learns complex patterns.
- Can automatically learn useful features.
- Performs well on many image, audio and language tasks.
- Supports end-to-end learning.

# Disadvantages:
- Often requires large amounts of data.
- Can require powerful hardware, such as GPUs.
- Training may take a long time.
- Can be difficult to interpret.
- May overfit training data.
